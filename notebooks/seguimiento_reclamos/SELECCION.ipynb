# Cobro indebido recurrente: selección para el LAB

Se prepara una prueba de conversación de segunda atención. El diálogo es **reconstruido**, no una transcripción. La selección demuestra que hay reclamos de cobro indebido de reclamantes recurrentes en proceso; no demuestra fallas de primera resolución del mismo incidente.

## Fuente y método

DuckDB local del EDA integral, tabla `complaints`, una fila por reclamo. Filtro `subcategory = Cobro indebido`, sin joins, sin datos de titulares en la salida. `is_repeat_complainer` se interpreta literalmente como recurrencia del reclamante. La consulta está en `seleccion.sql`; el archivo original permanece fuera de Git.

```python
from pathlib import Path
import duckdb
import pandas as pd
folder = Path.cwd()
root = folder.parents[1]
con = duckdb.connect(str(root / 'dataset/_eda_integral/integral.duckdb'), read_only=True)
sql = (folder / 'seleccion.sql').read_text(encoding='utf-8')
result = con.execute(sql).fetchdf()
con.close()
result.T.rename(columns={0:'Resultado'})
```

```python
row = result.iloc[0]
assert row['fee_cases'] == 12194
assert row['repeat_claimant_cases'] == 1865
assert row['in_process_repeat_claims'] > 0
assert row['description_variants'] == 1
assert row['declared_interaction_links'] == 0
print('Reconciliado con detalle_subcategories.csv del EDA; enlaces ausentes y descripción genérica confirmados.')
```

## Interpretación

Estos registros justifican probar continuidad de una revisión, pero no permiten reconstruir una conversación auténtica ni qué solución falló. Se redactó un caso de comisión con importe, folio y condiciones ficticios. No se une a productos ni movimientos del dataset.
